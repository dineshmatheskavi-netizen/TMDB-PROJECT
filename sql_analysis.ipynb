1)List all movies in the Action genre released after 2015, showing title and release date.

##JOIN (movies ⋈ movie_genres ⋈ genres)

```python
query = """SELECT 
           m.title,
           m.release_date,
           m.day_name
           FROM movies AS m
           JOIN movie_genres AS mg ON m.movie_id = mg.movie_id
           JOIN genres AS g ON mg.genre_id = g.genre_id
           WHERE g.genre_name = 'Action' AND YEAR(m.release_date) > 2015;"""

result = pd.read_sql(query, engine)

print(result)
```

2)Find the top 10 highest-grossing movies along with the genre(s) each belongs to.

JOIN + ORDER BY + LIMIT

```python
query = """SELECT
           m.title,
           m.revenue,
           GROUP_CONCAT(DISTINCT g.genre_name ORDER BY g.genre_name SEPARATOR ', ') AS genres
           FROM movies AS m
           JOIN movie_genres AS mg ON m.movie_id = mg.movie_id
           JOIN genres AS g ON mg.genre_id = g.genre_id
           GROUP BY m.movie_id, m.title, m.revenue
           ORDER BY m.revenue DESC LIMIT 10;"""

result = pd.read_sql(query, engine)

print(result)
```

3) Calculate the average budget and average revenue for each genre, sorted by average revenue descending

JOIN + GROUP BY + AVG()

```python
query = """SELECT g.genre_name AS genre,
           AVG(m.budget) AS avg_budget,
           AVG(m.revenue) AS avg_revenue
           FROM movies AS m
           JOIN movie_genres AS mg ON m.movie_id = mg.movie_id
           JOIN genres AS g ON mg.genre_id = g.genre_id
           GROUP BY g.genre_id, g.genre_name
           ORDER BY avg_revenue DESC;"""

result = pd.read_sql(query, engine)

print(result)
```

```python
4) List the top 10 actors who have appeared in the most movies in the dataset.(cast)

GROUP_BY + COUNT()

```

```python
query = """SELECT c.actor_name AS actor_name,
           COUNT(DISTINCT c.movie_id) AS total_movies
           FROM casting AS c
           GROUP BY c.person_id, c.actor_name
           ORDER BY total_movies DESC LIMIT 10;"""

result = pd.read_sql(query, engine)

print(result)
```

5) Find all directors who have directed more than 3 movies, along with their movie count.

GROUP BY + HAVING

```python
query = """SELECT person_id,
           person_name AS director_name,
           COUNT(DISTINCT movie_id) AS movie_count
           FROM crewing
           WHERE department = 'Directing' AND job = 'Director'
           GROUP BY person_id, person_name
           HAVING COUNT(DISTINCT movie_id) > 3
           ORDER BY movie_count DESC;"""

result = pd.read_sql(query, engine)

print(result)
```

6) List the top 10 most frequently used keywords across all movies.

GROUP BY + COUNT()

```python
query = """SELECT keyword_name AS keywords,
           COUNT(DISTINCT movie_id) AS movie_count
           FROM movie_keywords
           GROUP BY keyword_name
           ORDER BY movie_count DESC LIMIT 10;"""

result = pd.read_sql(query, engine)

print(result)
```

```python
7) Find movies where the budget was above the overall average budget but the revenue was below the overall average revenue.

Subquery comparison against AVG()
```

```python
query = """SELECT title,budget,revenue
           FROM movies
           WHERE budget > (SELECT AVG(budget)FROM movies) AND revenue < (SELECT AVG(revenue)FROM movies);"""

result = pd.read_sql(query, engine)

print(result)
```

8) List all actors who have appeared in a movie directed by a specific director (e.g., Christopher Nolan).

JOIN (cast ⋈ crew )

```python
query = """SELECT DISTINCT c.actor_name AS actor_name,c.person_id as person_id
           FROM casting AS c
           JOIN crewing AS cr ON c.movie_id = cr.movie_id
           WHERE cr.person_name = 'James Cameron' AND cr.job = 'Director' AND cr.department = 'Directing'
           ORDER BY actor_name;"""

result = pd.read_sql(query, engine)

print(result)
```

```python
9) Find the genre with the highest average rating, considering only genres with at least 20 movies.

JOIN + GROUP BY + HAVING + AVG()
```

```python
query = """SELECT g.genre_name AS genre_name,
           AVG(m.vote_average) AS avg_rating,
           COUNT(DISTINCT m.movie_id) AS total_movies
           FROM movies AS m
           JOIN movie_genres AS mg ON m.movie_id = mg.movie_id
           JOIN genres AS g ON mg.genre_id = g.genre_id
           GROUP BY g.genre_id, g.genre_name
           HAVING COUNT(DISTINCT m.movie_id) >= 20
           ORDER BY avg_rating DESC LIMIT 1;"""

result = pd.read_sql(query, engine)

print(result)
```

```python
10) List the top 10 movies with the highest number of keywords tagged to them. 

JOIN + GROUP BY
```

```python
query = """SELECT
           m.title,
           COUNT(DISTINCT k.keyword_name) AS total_keywords
           FROM movies AS m
           JOIN movie_keywords AS k ON m.movie_id = k.movie_id
           GROUP BY m.movie_id, m.title
           ORDER BY total_keywords DESC LIMIT 10;"""

result = pd.read_sql(query, engine)

print(result)
```
