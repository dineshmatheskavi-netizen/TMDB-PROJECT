```python
df_movies["release_date"]=pd.to_datetime(df_movies["release_date"],format="%Y-%m-%d")
```

```python
df_movies["year"]=df_movies["release_date"].dt.year
```

```python
df_movies["month"]=df_movies["release_date"].dt.month
```

```python
df_movies["day"]=df_movies["release_date"].dt.day
```

```python
df_movies["day_name"]=df_movies["release_date"].dt.day_name()
```

```python
df_movies
```

```python
df_movies.drop("status",axis=1,inplace=True)
```

```python
df_movies.drop("profit",axis=1,inplace=True)
```

```python
df_movies.duplicated().sum()
```

```python
df_movies=df_movies.drop_duplicates(keep="first")
```

```python
df_movies[df_movies["budget"]==0]
```

```python
import numpy as np

df_movies["budget"] = df_movies["budget"].replace(0, np.nan)
df_movies["revenue"] = df_movies["revenue"].replace(0, np.nan)
```

```python
df_movies[["budget", "revenue"]].isnull().sum()
```

```python
df_movies["profit"]=df_movies["revenue"]-df_movies["budget"]
```

```python
df_movies
```

```python
df_movies.to_csv("Cleaned_movies.csv",index=False)
```
