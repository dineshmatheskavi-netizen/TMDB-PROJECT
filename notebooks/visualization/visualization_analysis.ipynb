```python
df_budget_revenue = df_movies[(df_movies["budget"] > 0) & (df_movies["revenue"] > 0)]

fig = px.scatter(
    df_budget_revenue,
    x="budget",
    y="revenue",
    trendline="ols",
    hover_name="title",
    title="Movie vs Budget",
    labels = {"budget" : "Movie budget",
              "revenue" : "Movie revenue"}
)

fig.show()
```

VISUALIZATION CHART 6

insights - War genre movies got the highest avg rating 
           Horror genre movies got the lowest avg rating 

```python
df_rating_genre = df_movie_genres.merge(
    df_genres,
    on="genre_id",
    how="left"
)


df_rating_genre = df_rating_genre.merge(
    df_movies[["movie_id", "vote_average"]],
    on="movie_id",
    how="left"
)


df_rating_genre = df_rating_genre.dropna(
    subset=["vote_average", "genre_name"]
)


df_avg_rating = (
    df_rating_genre
    .groupby("genre_name")["vote_average"]
    .mean()
    .reset_index(name="avg_rating")
    .sort_values("avg_rating")
)


fig = px.scatter(
    df_avg_rating,
    x="avg_rating",
    y="genre_name",
    title="Average Rating by Genre — Dot Plot",
    labels={
        "avg_rating": "Average Rating",
        "genre_name": "Genre"
    },
    text="avg_rating"
)

fig.update_traces(
    marker=dict(size=12),
    texttemplate="%{text:.2f}",
    textposition="middle right"
)

fig.update_layout(
    xaxis_range=[0, 10]
)

fig.show()
```

VISUALIZATION CHART 7

insights - Revenue and budget has strong correlation
           Budget and vote_average has strong negative correlation

```python
df_corr = df_movies[
    [
        "budget",
        "revenue",
        "runtime",
        "popularity",
        "vote_average",
        "vote_count"
    ]
]

corr_matrix = df_corr.corr()

fig = px.imshow(
    corr_matrix,
    text_auto=".2f",
    color_continuous_scale="RdBu_r",
    zmin=-1,
    zmax=1,
    title="Correlation Heatmap of Movie Features"
)

fig.update_layout(
    width=800,
    height=700
)

fig.show()
```

VISUALIZATION CHART 8

insights - Almost all movie vote_average will be around 6 to 8

```python
df_primary_genre = (
    df_movie_genres
    .drop_duplicates(subset="movie_id", keep="first")
    .merge(df_genres, on="genre_id", how="left")
    [["movie_id", "genre_name"]]
    .rename(columns={"genre_name": "primary_genre"})
)

df_pairplot = df_movies.merge(
    df_primary_genre,
    on="movie_id",
    how="left"
)

cols = [
    "budget",
    "revenue",
    "popularity",
    "vote_average",
    "primary_genre"
]

df_pairplot = df_pairplot[cols].dropna()

fig = px.scatter_matrix(
    df_pairplot,
    dimensions=[
        "budget",
        "revenue",
        "popularity",
        "vote_average"
    ],
    color="primary_genre",
    title="Movie Pairplot by Primary Genre",
    opacity=0.5
)

fig.update_traces(
    diagonal_visible=True,
    showupperhalf=False,
    marker=dict(size=5)
)

fig.show()
```
