```python
6)List all movies whose runtime is a statistical outlier compared to the rest of the dataset.
```

```python
Q1=df_movies["runtime"].quantile(0.25)
Q3=df_movies["runtime"].quantile(0.75)

IQR = Q3 - Q1

df_movies_runtime_outliers=df_movies[(df_movies["runtime"] < Q1 - 1.5 * IQR) | 
                                     (df_movies["runtime"] > Q3 + 1.5 * IQR) ]

df_movies_runtime_outliers[["movie_id","title","runtime"]]
```

```python
7) List each movie's primary genre (the first genre listed) and compare average popularity across primary genres.
```

```python
primary_genres = (
    df_movie_genres
    .sort_values("genre_id")
    .drop_duplicates(subset="movie_id", keep="first")
    .merge(df_genres, on="genre_id", how="left")
    .merge(df_movies[["movie_id", "title", "popularity"]], on="movie_id", how="left")
)

primary_genres[["movie_id", "title", "genre_name", "popularity"]]
```

```python
primary_genres.groupby("genre_name")["popularity"].mean().sort_values(ascending=False)
```

```python
8) List the top 10 most prolific actors by number of movies, along with their average movie rating
   across those movies.
```

```python
top_10_actors = (
    df_cast
    .merge(
        df_movies[["movie_id", "vote_average"]],
        on="movie_id",
        how="left"
    )
    .groupby(["person_id", "actor_name"])
    .agg(
        movie_count=("movie_id", "nunique"),
        avg_rating=("vote_average", "mean")
    )
    .sort_values("movie_count", ascending=False)
    .head(10)
    .reset_index()
)

print(top_10_actors)
```

```python
10) List movies whose popularity score and vote count don't follow the general pattern between
    the two — i.e., unusually high popularity with a low vote count, or vice versa.
```

```python
df_movies.columns
```

```python
pop_high = df_movies["popularity"].quantile(0.75)
pop_low = df_movies["popularity"].quantile(0.25)

vote_high = df_movies["vote_count"].quantile(0.75)
vote_low = df_movies["vote_count"].quantile(0.25)

unusual_movies = df_movies[((df_movies["popularity"] >= pop_high) & (df_movies["vote_count"] <= vote_low)) |
                           ((df_movies["popularity"] <= pop_low) & (df_movies["vote_count"] >= vote_high))]


```

```python
unusual_movies[["title","popularity","vote_count"]].sort_values("popularity",ascending=False)
```
